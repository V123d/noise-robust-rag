```python
import pandas as pd
import urllib.request
import zipfile
import os

# 1. 设置下载链接和路径
url = "https://raw.githubusercontent.com/SophonPlus/ChineseNlpCorpus/master/datasets/online_shopping_10_cats/online_shopping_10_cats.zip"
zip_path = "online_shopping_10_cats.zip"
csv_path = "online_shopping_10_cats.csv"

# 2. 下载与解压逻辑
try:
    if not os.path.exists(csv_path):
        if not os.path.exists(zip_path):
            print("正在从 GitHub 下载 10_cats 数据集，请稍候...")
            urllib.request.urlretrieve(url, zip_path)
            print("✅ 下载压缩包完成！")
        
        print("正在解压文件...")
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall()
        print("✅ 解压完成！")
    else:
        print("✅ CSV数据文件已存在，跳过下载和解压。")

    # 3. 使用 Pandas 加载数据
    print("正在加载数据到 Pandas DataFrame...")
    df = pd.read_csv(csv_path)

    print(f"\n🎉 成功！共获取 {len(df)} 条中文电商评论。")
    
    # 4. 探查数据长什么样
    print("\n--- 数据集前 5 行预览 ---")
    display(df.head())
    
    print("\n--- 包含的商品类别分布 ---")
    display(df['cat'].value_counts())

except Exception as e:
    print(f"❌ 下载或处理时出现错误：{e}")
    print("提示：如果是因为网络问题无法连接 Github，请手动从网页下载该 zip 文件，解压出 csv 文件放到当前目录下，然后直接运行 df = pd.read_csv('online_shopping_10_cats.csv') 即可。")
```

```python
import pandas as pd

# 1. 读取刚才保存好的原始 CSV 文件
csv_path = "online_shopping_10_cats.csv"
df_all = pd.read_csv(csv_path)

# 2. 选取一个垂直领域（这里我们选择“平板”，因为电子产品非常适合做事实一致性检验）
target_category = "平板"
df_tablet = df_all[df_all['cat'] == target_category].copy()

# 3. 数据清洗：去除空评论，去除极短的无意义评论（少于5个字的）
df_tablet.dropna(subset=['review'], inplace=True)
df_tablet = df_tablet[df_tablet['review'].apply(lambda x: len(str(x)) > 5)]

# 4. 重置索引
df_tablet.reset_index(drop=True, inplace=True)

print(f"✅ 成功提取【{target_category}】类别的评论！清洗后共剩余 {len(df_tablet)} 条数据。")

# 5. 保存为我们实验专属的知识库文件
corpus_path = "tablet_corpus.csv"
df_tablet.to_csv(corpus_path, index=False, encoding='utf-8-sig')
print(f"✅ 专属知识库已保存为: {corpus_path}")

# 随便抽 3 条含有明显“无关噪声”或“主观情绪”的评论看看
print("\n--- 真实的带噪评论示例 ---")
for i, review in enumerate(df_tablet['review'].sample(3, random_state=42)):
    print(f"评论 {i+1}: {review}")
```
